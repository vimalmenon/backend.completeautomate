## Add YouTube Channel Task

```python
from backend.data import Task, YouTubeChannelJobData
from backend.enum.job import JobEnum
from backend.enum.team import TeamEnum
from backend.enum.status import TaskStatusEnum
from uuid import uuid4
from datetime import datetime
from backend.config.env import env
from backend.database import TaskDB


job_data = YouTubeChannelJobData(
    channel_id=env.YOUTUBE_CHANNEL_ID
)

task = Task(
    id=uuid4(),
    job_type=JobEnum.YouTubeChannel,
    created_by=TeamEnum.OWNER,
    status=TaskStatusEnum.IN_PROGRESS,
    payload=job_data.to_json(),
    created_at=datetime.now(),
	trail=[]
)

print(task)

if safety_check:
	TaskDB().add_task(task)
	print(f"Task added to database with ID: {task.id}")
else:
	print("Task not added to database as safety check is set to False.")
```

## Get Class from S3 path

```python
# Validate nested key parsing from S3 path
parsed = S3Data.to_cls_from_path("json/test/json/file.json")

assert parsed.key == "test/json", f"Unexpected key: {parsed.key}"
assert parsed.s3_key == "json/test/json/file.json", f"Unexpected s3_key: {parsed.s3_key}"
print("OK:", parsed.key, parsed.s3_key)
```
